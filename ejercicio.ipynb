```python
import numpy as np
import statistics as stats
import matplotlib.pyplot as plt
# Pesos de las zanahorias en gramos 🥕
zanahorias = [300, 280, 290, 310, 275, 290, 295, 315, 290, 280, 310, 305]
# Estadísticas descriptivas
media = np.mean(zanahorias)
mediana = np.median(zanahorias)
moda = stats.mode(zanahorias)
desviacion = np.std(zanahorias)
# Resultados
print(f"Media: {media} g")
print(f"Mediana: {mediana} g")
print(f"Moda: {moda} g")
print(f"Desviación estándar: {desviacion:.2f} g")
# Crear eje X (índice de cada zanahoria)
x = list(range(1, len(zanahorias) + 1))
y = zanahorias
# Gráfico
#plt.plot(x, y, marker='o') Este es grafico de lineas
plt.bar(x,y) #Este es el grafico de barra
plt.title("Peso de las zanahorias")
plt.xlabel("Número de zanahoria")
plt.ylabel("Peso (g)")
plt.grid()
plt.show()
```
