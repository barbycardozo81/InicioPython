```python
import random

def juego_adivinanza():
    numero_secreto = random.randint(1, 100)
    intentos = 0
    adivinando = False

    print("¡Bienvenido al juego de adivinanza!")
    print("Estoy pensando en un número entre 1 y 100.")

    while not adivinando:
        experimento = int(input("Adivina el número: "))
        intentos += 1

        if experimento < numero_secreto:
            print("El número que estoy pensando es mayor.")
        elif experimento > numero_secreto:
            print("El número que estoy pensando es menor.")
        else:
            adivinando = True
            print(f"¡Felicidades! Adivinaste el número {numero_secreto} en {intentos} intentos.")

if __name__ == "__main__":
    juego_adivinanza()
```
