Aprenderemos conceptos basicos detras de la regresión lineal y como implementarla en python utilizando bibliotecas basicas como: Pandas, NumPy, Matplotlib, Scikitlearn

```python
#Importamos las librerias necesarias
import pandas as pd #Manipulación de datos
import numpy as np #Calculos numericos
import matplotlib.pyplot as plt #Graficos
from sklearn.linear_model import LinearRegression
import warnings #para gestionar advertencias

#ignoramos los warnings
warnings.filterwarnings('ignore')

#Utilizaremos un conjunto de datos que relaciona la experiencia laboral con e
#con el salario anual

#Leeremos el archivo CSV y cargaremos los datos en un DataFrame

df = pd.read_csv("Salary.csv")

#Mostraremos por pantalla las primeras 6 filas
df.head(6)
```

Graficamos los datos para observar la dispersión y correlación entre las variables.

```python
plt.scatter(df["YearsExperience"], df["Salary"])

plt.xlabel("Years of Experience") #Nombre de la etiqueta que va por el borde
plt.ylabel("Salary")              #Eje y
plt.title("Salary vs Experience") #Eje X
plt.show()                        #Mostramos el grafico
```

```python
#Verificamos si la correlación es positiva
#Creamos un objeto de modelo de regreción lineal
modelo = LinearRegression()

#Ajustamos el modelo a los datos

modelo.fit(df[["YearsExperience"]], df["Salary"])

prediccion = modelo.predict([[15]])

print(f"Predicción del salario anual para 15 años de experiencia {prediccion}")
```

```python
#Grafico de dispersión de los ejes
plt.scatter(df["YearsExperience"], df["Salary"])

plt.xlabel("Years of experience")
plt.ylabel("Salary")
#Superponer una linea de regresión del modelo en el grafico.

plt.plot(df["YearsExperience"], modelo.predict(df[["YearsExperience"]]), color="red")

#Grafico
plt.show()
```

```python
print("R^2: {}".format(modelo.score(df[["YearsExperience"]], df["Salary"])))
```
